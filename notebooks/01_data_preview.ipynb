# A first look at the data
Run the demo commands in the README first. We will check how many records survived cleaning, how long the student histories are, and how many risk examples we have. The default input is made-up data; switch RUN_DIR to work/ednet after preparing EdNet.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd() if (Path.cwd() / 'scripts').exists() else Path.cwd().parent
RUN_DIR = ROOT / 'work' / 'demo'
summary = json.loads((RUN_DIR / 'summary.json').read_text(encoding='utf-8'))
print(summary['config'])
print(summary['cleaning'])
print(summary['risk_observability'])
```

```python
students = pd.read_csv(RUN_DIR / 'students.csv')
risk = pd.read_csv(RUN_DIR / 'risk_examples.csv')
display(students.head())
display(risk.head())
display(students.groupby('split')['n_interactions'].agg(['count', 'median', 'min', 'max']))
display(risk.groupby('split')['risk_label'].agg(['count', 'mean']))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 3))
students['n_interactions'].hist(ax=axes[0], bins=15)
axes[0].set_title('Answers per student')
risk['future_accuracy'].hist(ax=axes[1], bins=10)
axes[1].axvline(summary['config']['risk_threshold'], color='red', linestyle='--')
axes[1].set_title('Accuracy in the next 20 questions')
plt.tight_layout()
```
